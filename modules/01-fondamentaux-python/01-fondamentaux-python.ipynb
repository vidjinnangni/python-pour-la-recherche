## 2. Structures de données

Une structure de données est une manière d'organiser et de stocker plusieurs valeurs afin de pouvoir les manipuler plus facilement. Jusqu'à présent, nous avons vu qu'une variable permet de stocker une valeur. Mais en recherche, nous ne travaillons pas avec une seule valeur. À cet effet, Python propose différentes structures pour organiser ces données. Les principales structures que nous allons rencontrer sont :

- la liste (`list`) : une collection ordonnée et modifiable de valeurs ;
- le tuple (`tuple`) : une collection ordonnée, mais non modifiable ;
- le dictionnaire (`dict`) : une collection de données organisées sous forme de paires clé-valeur ;
- l'ensemble (`set`) : une collection de valeurs uniques.

Pour la recherche en sciences de gestion, ces structures sont particulièrement utiles pour **organiser, manipuler et préparer les données** avant leur analyse.

### 2.3. Les tuples

Un **tuple** est une collection ordonnée de valeurs, comme une liste. La principale différence est qu'un tuple est immuable : une fois créé, on ne peut plus modifier, ajouter ou supprimer ses éléments.

On crée généralement un tuple avec des parenthèses :

````python
coordonnees = (48.85, 2.35)  # latitude, longitude de Paris
````

Ici, le tuple contient deux valeurs :

```
48.85 → latitude
2.35  → longitude
```

On peut accéder à ses éléments avec leur index, exactement comme avec une liste :

````python
print(coordonnees[0])
print(coordonnees[1])
````

La différence avec une liste

Avec une liste, on peut modifier un élément :

```python
coordonnees_liste = [48.85, 2.35]

coordonnees_liste[0] = 50

print(coordonnees_liste)
```

Résultat :

```
[50, 2.35]
```

Avec un tuple, cette modification n'est pas possible :

````python
coordonnees = (48.85, 2.35)

coordonnees[0] = 50
````

Python renvoie alors une erreur `TypeError` : *'tuple' object does not support item assignment*

#### Quand utiliser un tuple ?

Un tuple est utile lorsque nous avons plusieurs valeurs qui doivent rester ensemble et ne doivent pas être modifiées. Les coordonnées d'un point géographique constituent un bon exemple : la latitude et la longitude forment un ensemble de valeurs qui peut être traité comme une seule information. On peut également utiliser un tuple pour représenter une paire de valeurs ou une date.

> **En pratique :** si vous avez besoin de modifier votre collection, utilisez généralement une liste. Si les valeurs doivent rester inchangées, un tuple peut être plus approprié.

## 3. Conditions

### 3.1. Définition

Les conditions permettent d'exécuter du code seulement si une situation donnée est vraie. Jusqu'à présent, Python exécutait les instructions que nous lui donnions. Avec les conditions, nous pouvons lui demander d'exécuter une instruction *seulement si une certaine situation est vraie*.

Par exemple :

````python
age = 20

if age >= 18:
    print("La personne est majeure.")
````

Ici, Python vérifie si age est supérieur ou égal à 18. Si la condition est vraie, il exécute l'instruction.

On peut également prévoir plusieurs situations :

````python
satisfaction = 2

if satisfaction >= 4:
    print("Client satisfait")
elif satisfaction == 3:
    print("Client neutre")
else:
    print("Client insatisfait")
````

Ici, on assigne la valeur 2 à la variable `satisfaction`. Ensuite, on écrit une condition pour tester cette valeur afin d'afficher un message :

- si `satisfaction` est supérieure ou égale à 4, le code affiche « Client satisfait » ;
- si `satisfaction` vaut 3, le code affiche « Client neutre » ;
- sinon, il affiche « Client insatisfait ».

Première condition (`if`) : elle vérifie si satisfaction >= 4. Comme satisfaction vaut 2, c'est faux.

Deuxième condition (`elif`) : elle vérifie si satisfaction == 3. C'est faux aussi

Alternative finale (`else`) : comme les autres tests sont faux, le code exécute le dernier bloc et affiche Client insatisfait.

### 3.2. Opérateurs de comparaison et logiques

Pour écrire des conditions, Python met à notre disposition deux catégories d'opérateurs : **les opérateurs de comparaison** et **les opérateurs logiques**.

#### 3.2.1. Les opérateurs de comparaison

Les opérateurs de comparaison permettent de comparer deux valeurs. Le résultat est toujours `True` ou `False`.

| Opérateur | Signification       | Exemple     |
| --------- | ------------------- | ----------- |
| `==`      | égal à              | `age == 18` |
| `!=`      | différent de        | `age != 18` |
| `>`       | supérieur à         | `age > 18`  |
| `<`       | inférieur à         | `age < 18`  |
| `>=`      | supérieur ou égal à | `age >= 18` |
| `<=`      | inférieur ou égal à | `age <= 18` |

#### 3.2.2. Les opérateurs logiques

Les opérateurs logiques permettent de combiner plusieurs conditions. Python en utilise principalement trois :

| Opérateur | Signification        | Exemple                   |
| --------- | -------------------- | ------------------------- |
| `and`     | et                   | `age >= 18 and age <= 65` |
| `or`      | ou                   | `age < 18 or age > 65`    |
| `not`     | inverse la condition | `not majeur`              |

Ces opérateurs sont particulièrement utiles pour sélectionner ou catégoriser des observations.

````python
secteur = "Technologies"
effectifs = 15

if secteur == "Technologies" and effectifs < 50:
    print("Start-up technologique")
````

> **Attention !** Il ne faut surtout pas confondre les opérateurs `=` et `==`. L'opérateur `=` sert à affecter une valeur à une variable. C'est un opérateur d'affectation. Tandis que l'opérateur `==` sert à comparer deux valeurs. C'est un opérateur de comparaison d'égalité.

## 4. Les boucles

En programmation, on est amené à effectuer la même opération sur plusieurs valeurs. Plutôt que de répéter le même code manuellement, on peut utiliser **une boucle**. Elle va nous permettre de répéter automatiquement une ou plusieurs instructions plusieurs fois.

Par exemple, si nous voulons afficher plusieurs âges :

````python
ages = [25, 32, 28, 41]

for age in ages:
    print(age)
````

Python parcourt alors chaque valeur de la liste et exécute l'instruction `print()`.

Il existe principalement deux types de boucles en Python :

- `for` : permet de parcourir une collection de valeurs ou de répéter une action un nombre déterminé de fois ;
- `while` : permet de répéter une action tant qu'une condition est vraie.

Les boucles sont particulièrement utiles en recherche pour **parcourir des données, automatiser des traitements répétitifs et appliquer une même opération à plusieurs observations ou fichiers**.

### 4.1. La boucle `for`

La boucle `for` permet de répéter une instruction pour chaque élément d'une collection ou d'une séquence.

````python
secteurs = ["Industrie", "Commerce", "Technologies"]

for secteur in secteurs:
    print("Secteur :", secteur)
````

````python
entreprise = {"nom": "Acme SA", "secteur": "Industrie", "effectifs": 120}

for cle, valeur in entreprise.items():
    print(cle, "->", valeur)
````

On peut aussi utiliser `for` avec `range()` pour répéter une action un nombre déterminé de fois :

````python
for i in range(5):
    print(i)
````

> Une boucle `for` peut notamment permettre de parcourir les observations d'une enquête, plusieurs fichiers de données ou différents éléments d'un corpus afin d'appliquer le même traitement à chacun.

### 4.2. La boucle `while`

La boucle `while` répète une ou plusieurs instructions tant qu'une condition est vraie. Elle est moins utilisée que `for` en pratique, mais reste utile quand on ne connaît pas à l'avance le nombre d'itérations.

Exemple :

````python
i = 0

while i < 5:
    print(i)
    i += 1
````

Tant que `i < 5` est vrai, Python exécute le contenu de la boucle.

````python
effectifs = 10
annees = 0

while effectifs < 50:
    effectifs = effectifs * 1.1  # croissance de 10% par an
    annees += 1

print(f"Il faut {annees} ans pour dépasser 50 salariés (effectifs = {effectifs:.1f})")
````

### 4.3. Compréhensions de liste

Une compréhension de liste est une manière concise de créer une nouvelle liste à partir d'une collection de valeurs. Elle permet notamment de remplacer certaines boucles `for` simples par une écriture plus courte.

Par exemple, avec une boucle classique :

````python
ages = [18, 20, 25, 30]

ages_majeurs = []

for age in ages:
    ages_majeurs.append(age)
````

Avec une compréhension de liste :

````python
ages = [18, 20, 25, 30]

ages_majeurs = [age for age in ages]
````

On peut aussi appliquer une transformation à chaque élément :

````python
ages = [18, 20, 25, 30]

ages_dans_5_ans = [age + 5 for age in ages]

print(ages_dans_5_ans)
````

Il est aussi possible d'ajouter une condition :

````python
ages = [16, 18, 20, 25, 30]

ages_majeurs = [age for age in ages if age >= 18]

print(ages_majeurs)
````

La structure générale est :

```python
[nouvelle_valeur for élément in collection if condition]
```

> Une compréhension de liste permet de créer rapidement une nouvelle liste en parcourant une collection, avec éventuellement une transformation ou une condition. En recherche, elles peuvent être utiles pour transformer ou filtrer rapidement des données avant de passer à des outils plus spécialisés comme `pandas`.

````python
effectifs_entreprises = [12, 45, 230, 8, 95]

# Version "classique" avec une boucle for
grandes_entreprises = []
for e in effectifs_entreprises:
    if e > 50:
        grandes_entreprises.append(e)
print(grandes_entreprises)

# Même résultat en une ligne, avec une compréhension de liste
grandes_entreprises_v2 = [e for e in effectifs_entreprises if e > 50]
print(grandes_entreprises_v2)
````

## 5. Les fonctions

Une fonction est un bloc de code qui réalise une tâche précise et que l'on peut réutiliser plusieurs fois dans un programme. Elle prend des **paramètres** en entrée et peut **retourner** un résultat.

Les fonctions permettent notamment d'éviter de répéter le même code et de mieux organiser un programme.

Exemple :


````python
def saluer():
    print("Bonjour !")

# On peut ensuite appeler la fonction autant de fois que nécessaire :

saluer()
saluer()
````

### 5.1. Les paramètres

Une fonction peut recevoir des informations appelées **paramètres** :

````python
def saluer(nom):
    print("Bonjour", nom)

saluer("Alice")
saluer("Paul")
````

Ici, `nom` est un paramètre de la fonction.

### 5.2. Retourner une valeur

Une fonction peut également produire un résultat avec `return` :

````python
def additionner(a, b):
    return a + b

resultat = additionner(10, 5)
print(resultat)
````

La structure générale est donc :

```python
def nom_de_la_fonction(paramètres):
    # instructions
    return résultat
```

> Une fonction permet de regrouper des instructions pour réaliser une tâche précise et réutilisable. Les fonctions sont particulièrement utiles pour **automatiser et standardiser les traitements de données**.

Par exemple, si l'on doit calculer le taux de croissance de nombreuses entreprises, on peut créer une fonction une seule fois :

```python
def taux_croissance(ancien, nouveau):
    return (nouveau - ancien) / ancien * 100
```

Puis la réutiliser pour différentes observations.

````python
def moyenne_ponderee(valeurs, poids):
    total_pondere = sum(v * p for v, p in zip(valeurs, poids))
    total_poids = sum(poids)
    return total_pondere / total_poids

notes_satisfaction = [4, 5, 2, 3]
poids_repondants = [1, 2, 1, 1]  # certains répondants comptent double, par exemple

resultat = moyenne_ponderee(notes_satisfaction, poids_repondants)
print(f"Moyenne pondérée : {resultat:.2f}")
````

### 5.3. Pourquoi découper son code en fonctions ?

- **Réutilisabilité** : écrire une fois, utiliser partout
- **Lisibilité** : un nom de fonction explicite (`moyenne_ponderee`) documente ce que fait le code, sans avoir à relire les détails
- **Facilité de test et de débogage** : plus simple de vérifier qu'un petit bloc isolé fonctionne correctement

## 6. Erreurs et débogage

Vous allez rencontrer des erreurs très régulièrement. C'est normal, même pour des programmeurs expérimentés ! Une erreur survient lorsque Python ne peut pas exécuter le code comme prévu. Le débogage consiste à identifier et à corriger la cause de cette erreur.

Les erreurs font partie intégrante de la programmation. Écrire du code, c'est aussi apprendre à comprendre les messages d'erreur.


````python
# Exécutez cette cellule et lisez attentivement le message d'erreur (traceback)
print(variable_qui_nexiste_pas)
````

Ce type d'erreur est un `NameError` : Python vous dit qu'il ne connaît pas de variable portant ce nom. Le traceback indique la ligne exacte où l'erreur s'est produite ; c'est la première chose à regarder.

### 6.1. Erreurs courantes à reconnaître

| Erreur | Cause typique |
|---|---|
| `NameError` | Utiliser une variable qui n'a pas été définie (ou faute de frappe dans son nom) |
| `TypeError` | Mélanger des types incompatibles (ex. additionner un texte et un nombre) |
| `IndexError` | Accéder à un index de liste qui n'existe pas (ex. `liste[10]` sur une liste de 5 éléments) |
| `KeyError` | Accéder à une clé de dictionnaire qui n'existe pas |


On peut distinguer trois situations courantes :

#### 6.1.1. Les erreurs de syntaxe

Elles surviennent lorsque le code ne respecte pas les règles d'écriture de Python.

````python
if age >= 18
    print("Majeur")
````

Python signale alors une erreur parce qu'il manque `:` après la condition.

#### 6.1.2. Les erreurs d'exécution

Le code est correctement écrit, mais une erreur survient pendant son exécution.

````python
age = int("vingt")
````

Python ne peut pas convertir "vingt" en nombre entier.

#### 6.1.3. Les erreurs logiques

Le programme s'exécute sans afficher d'erreur, mais le résultat obtenu est incorrect.

````python
prix = 100
reduction = 20

prix_final = prix + reduction
````

Le code fonctionne, mais si `reduction` représente une réduction de 20 %, le calcul est incorrect.

## 6.2. Comment déboguer ?

Face à une erreur, il faut prendre l'habitude de :

- lire le message d'erreur ;
- identifier la ligne concernée ;
- comprendre ce que Python attendait ;
- vérifier les variables et les valeurs utilisées ;
- corriger le code et tester à nouveau.

Python dispose d'une bibliothèque standard riche et l'écosystème externe va beaucoup plus loin.

### 7.1. Importer une bibliothèque

Pour utiliser une bibliothèque, on l'importe généralement avec import :


```python
import math
```

On peut ensuite accéder à ses fonctionnalités :

```python
math.sqrt(25)
math.pi
```

On peut aussi importer une fonctionnalité spécifique d'une bibliothèque :

```python
from math import sqrt

sqrt(25)
```

````python
import math

print(math.sqrt(16))  # racine carrée, depuis la bibliothèque standard

# Alias : convention très répandue pour les bibliothèques externes
import numpy as np

print(np.mean([10, 20, 30]))  # moyenne, depuis numpy
````

`math` fait partie de la bibliothèque standard de Python (toujours disponible, aucune installation nécessaire). `numpy`, en revanche, est une bibliothèque externe. Elle doit être installée (c'est fait via `requirements.txt`, voir le README principal) avant de pouvoir être importée.

> Le `as np` crée simplement un alias, c'est-à-dire un nom court permettant d'utiliser la bibliothèque plus facilement. C'est un usage standard et reconnu de tous. Cela évite d'écrire le nom entier de la biblithèque à chaque appel de fonction ou de méthode. Cela rend aussi les lignes de code plus courtes et moins chargées visuellement. Enfin, cet usage permet à n'importe quel autre programmeur de reconnaître immédiatement la bibliothèque utilisée.

### 7.2. Les bibliothèques pour la recherche

Python possède un vaste écosystème adapté à la recherche et à l'analyse de données.

Quelques exemples :

| Bibliothèque   | Utilisation                         |
| -------------- | ----------------------------------- |
| `NumPy`        | Calcul numérique                    |
| `pandas`       | Manipulation et analyse de données  |
| `Matplotlib`   | Visualisation des données           |
| `Seaborn`      | Visualisation statistique           |
| `SciPy`        | Calcul scientifique et statistiques |
| `statsmodels`  | Modèles statistiques et régressions |
| `scikit-learn` | Machine learning                    |
