## 1. Découverte du DataFrame

### 1.1. Qu'est-ce que `pandas` ?

`pandas` est une bibliothèque Python spécialisée dans **la manipulation et l'analyse des données**. Elle est particulièrement adaptée aux données organisées sous forme de tableaux, comme celles que l'on retrouve dans les fichiers Excel, CSV ou dans les bases de données. Imaginez `pandas` comme un tableau Excel, mais que vous pilotez avec du code plutôt qu'à la souris. C'est l'outil que vous utiliserez pour la quasi-totalité de vos traitements de données de recherche, quel que soit votre domaine en sciences de gestion. Elle permet notamment de :

- importer des fichiers CSV ou Excel ;
- explorer une base de données ;
- sélectionner et filtrer des observations ;
- traiter les valeurs manquantes ;
- créer ou modifier des variables ;
- regrouper et agréger des données ;
- calculer des statistiques descriptives ;
- préparer les données pour des analyses statistiques et des visualisations.

Pour utiliser `pandas`, il faut l'importer sous l'alias `pd`

```python
import pandas as pd
```

On peut ensuite utiliser ses principales fonctionnalités avec `pd`. Par exemple, pour créer un petit tableau :


````python
import pandas as pd

donnees = {
    "age": [25, 32, 28],
    "satisfaction": [4, 3, 5]
}

df = pd.DataFrame(donnees)
````

`df` (la variable) contient alors un `DataFrame`, c'est-à-dire une structure de données tabulaire composée de lignes et de colonnes.

Avec Pandas, les données sont principalement organisées à l'aide de deux structures : les **Series** et les **DataFrames**. Une Series représente une colonne de données, tandis qu'un DataFrame représente un tableau composé de plusieurs colonnes.

### 1.2. Series

Une Series est une suite de valeurs organisée avec un index.

````python
import pandas as pd

ages = pd.Series([25, 32, 28, 41])
````

On peut la représenter ainsi :

```
0    25
1    32
2    28
3    41
```

Les nombres à gauche constituent l'index, tandis que les valeurs à droite correspondent aux données. Une Series peut représenter une variable d'une enquête, par exemple l'âge des répondants.

### 1.3. DataFrame

Un DataFrame est un tableau de données composé de lignes et de colonnes. C'est l'équivalent d'une feuille Excel.

````python
donnees = {
    "age": [25, 32, 28],
    "satisfaction": [4, 3, 5]
}

df = pd.DataFrame(donnees)
````

On obtient :

```
   age  satisfaction
0   25             4
1   32             3
2   28             5
```

Chaque colonne est une Series et l'ensemble des colonnes forme le DataFrame.

On peut donc retenir :

- **Series** = une colonne de tableau
- **DataFrame** = un tableau entier de données

> Vous verrez ces deux mots partout dans la documentation `pandas` et dans les messages d'erreur. Cette distinction est importante pour la suite du cours, car la plupart des opérations de manipulation et d'analyse de données avec Pandas se feront sur des DataFrames.

## 2. Sélectionner et filtrer

Vous savez désormais comment charger des données à partir d'un fichier CSV ou Excel dans un DataFrame. Maintenant, il peut arriver que vous n'ayez besoin que d'une partie des données chargées. Dans ce cas, on va utiliser les options de **sélection** et de **filtrage**. La sélection consiste à choisir certaines lignes ou colonnes d'un DataFrame, tandis que le filtrage consiste à conserver uniquement les observations qui répondent à une condition.

### 2.1. Sélectionner une colonne

Pour extraire une seule colonne (une `Series` donc), on va utiliser des crochets avec le nom de la colonne entre guillemets :


````python
df["secteur"] # On obtient alors une Series contenant uniquement la variable secteur.
````

Notez la ligne tout en bas de l'affichage : `Name: secteur`, `Length: 404`, `dtype: str`. C'est la signature d'une `Series`, pas d'un `DataFrame`. Si on veut sélectionner **plusieurs colonnes** (et donc obtenir un `DataFrame`, pas une `Series`), on va plutôt utiliser une **liste** de noms de colonnes entre doubles crochets :


````python
df[["entreprise_id", "annee", "secteur", "teletravail"]]
````

***Remarque sur la syntaxe :** le premier `[ ]` sert à "accéder à des colonnes", le second `[ ]` (à l'intérieur) est la liste Python des noms voulus ; d'où les doubles crochets `[[...]]`. Cela peut prêter à confusion au début.*

### 2.2. `.iloc` et `.loc` : sélectionner des lignes

Pour sélectionner précisément des données (des lignes) dans un DataFrame, `pandas` propose deux principales méthodes :

- **`.iloc`** : permet de sélectionner des données à partir de leur **position numérique** (comme un index de liste 0, 1, 2...) ;
- **`.loc`** : permet de sélectionner des données à partir de leurs **étiquettes** (noms d'index ou de colonnes).


````python
# .iloc : "je veux la ligne à la position 0" (la toute première, peu importe son label)

df.iloc[0]
````

````python
# .loc : ici, comme l'index de notre DataFrame est numéroté 0, 1, 2... par défaut
# (pas de colonne définie explicitement comme index), .loc[0] donne le même résultat
# que .iloc[0] -- mais ce n'est PAS toujours le cas, voir la remarque juste après

df.loc[0]
````

**Pourquoi cette distinction ?** Si vous filtrez ou triez votre DataFrame, les labels des lignes (l'index) ne correspondent plus à leur position. Exemple :


````python
df_trie = df.sort_values("chiffre_affaires_keur", ascending=False)

# La ligne ayant le label (l'index) 42, quelle que soit sa position actuelle après le tri :
# df_trie.loc[42]

# La ligne qui est maintenant en première position après le tri, peu importe son label :

premiere_ligne_apres_tri = df_trie.iloc[0]
print(premiere_ligne_apres_tri[["entreprise_id", "annee", "chiffre_affaires_keur"]])
````

**Règle pratique** : Dans ce cours, nous utiliserons `.loc` presque tout le temps (pour sélectionner par condition, ce qu'on voit juste en dessous), et `.iloc` seulement quand nous voulons explicitement "la Nème ligne", indépendamment de son label.

### 2.3. Sélectionner lignes ET colonnes avec `.loc`

`.loc[lignes, colonnes]` 

> Remarquez la virgule qui sépare les deux :


````python
# Toutes les lignes, seulement 3 colonnes précises

df.loc[:, ["entreprise_id", "annee", "chiffre_affaires_keur"]]
````

Le `:` seul signifie "toutes les lignes". C'est la même logique que le *slicing* sur les listes Python vu au module 01.

### 2.4. Filtrer selon une condition

C'est probablement l'opération que vous ferez le plus souvent avec `pandas` : garder seulement les lignes qui respectent une condition. Le filtrage est très courant en recherche. Il permet par exemple de :

- sélectionner les répondants d'une catégorie donnée ;
- étudier uniquement les entreprises d'un secteur ;
- analyser une période particulière ;
- exclure certaines observations ;
- travailler sur un sous-échantillon.

La mécanique se fait en deux temps, même si on l'écrit généralement en une seule ligne.


````python
# Étape 1 : construire un "masque" -- une Series de True/False, une valeur par ligne

masque = df["teletravail"] == 1
masque.head()
````

**Ce qui se passe ici** : `pandas` compare `df["teletravail"]` (une colonne entière) à `1`, ligne par ligne, et retourne une Series de `True`/`False` de la même longueur que `df`. C'est ce qu'on appelle un **masque booléen**.


````python
# Étape 2 : utiliser ce masque pour filtrer le DataFrame -- il ne garde que les lignes où le masque vaut True

entreprises_teletravail = df[masque]
entreprises_teletravail.head()
````

En pratique, on combine généralement les deux mécaniques en une seule ligne :


````python
entreprises_teletravail = df[df["teletravail"] == 1]
print(f"{len(entreprises_teletravail)} lignes sur {len(df)} concernent des entreprises en télétravail")
````

### 2.5. Combiner plusieurs conditions

Pour combiner plusieurs conditions dans Pandas, on utilise les symboles :

- `&` pour "et" : les deux conditions doivent être vraies en même temps ;
- `|` pour "ou" : au moins une des conditions doit être vraie.

> [!NOTE]
> On n'utilise pas les opérateurs logiques `and` ("et") et `or` ("ou") de Python natif pour combiner des conditions portant sur des Series Pandas. De plus, chaque condition doit être placée entre parenthèses. Elles permettent à Pandas d'évaluer correctement chaque condition avant de les combiner.


````python
# Entreprises en télétravail ET dans le secteur Technologies

filtre = df[(df["teletravail"] == 1) & (df["secteur"] == "Technologies")]
filtre.head()
````

> [!IMPORTANT]
> Il faut faire attention à ne pas oublier les parenthèses autour de chaque condition. En effetn, écrire `df[df["teletravail"] == 1 & df["secteur"] == "Technologies"]` (sans les parenthèses) provoque une erreur, à cause de l'ordre de priorité des opérateurs en Python. Prenez le réflexe de **toujours mettre entre paranthèses chaque condition** quand vous en combinez plusieurs.

### 2.6. Trier les données

Il peut être aussi utile d'ordonner les données (après sélection et filtrage) selon une ou plusieurs variables. Avec `pandas`, le tri permet donc de réorganiser les lignes d'un DataFrame selon les valeurs d'une ou plusieurs colonnes. À cet effet, on utilise principalement la méthode `sort_values()`.


````python
# Les 5 entreprises (lignes) avec le chiffre d'affaires le plus élevé

df.sort_values("chiffre_affaires_keur", ascending=False).head()
````

**Décomposition de la commande** 

- `df.sort_values(...)` : Trie les lignes du DataFrame `df` en fonction d'une colonne spécifique.
- `"chiffre_affaires_keur"` : Désigne la colonne utilisée pour le tri (le chiffre d'affaires en milliers d'euros).
- `ascending=False` : Active le tri décroissant (du plus grand chiffre d'affaires au plus petit).
- `.head()` : Conserve uniquement les premières lignes du résultat (5 par défaut).

## 3. Nettoyer les données

Un jeu de données réel n'est presque jamais parfaitement propre. Avant toute analyse, il faut systématiquement vérifier : valeurs manquantes, doublons, types incohérents. On procède donc au nettoyage de données qui consiste à **identifier et corriger les problèmes qui peuvent affecter la qualité ou l'analyse des données**. Il s'agit de préparer les données pour qu'elles puissent être analysées de manière fiable.

Pour cela, on va commencer par vérifier les données avec `df.info()` pour connaître les types de variables et repérer certaines valeurs manquantes. On peut également rechercher les doublons avec `df.duplicated()` et compter les valeurs manquantes avec `df.isna().sum()`.

### 3.1. Repérer les valeurs manquantes

Une valeur manquante est une donnée qui n'a pas été renseignée ou qui est absente. La méthode `.isna()` permet de repérer ces valeurs. Elle retourne `True` lorsqu'une valeur est manquante et `False` lorsqu'elle est présente. Par exemple, si notre DataFrame contient :

```
    age   salaire
0    25    150000
1   NaN     200000
2    32       NaN
```

`df.isna()` retourne :

```
     age  salaire
0  False    False
1   True    False
2  False     True
```

> [!NOTE]
> NaN (Not a Number) est l'une des représentations utilisées par Pandas pour les valeurs manquantes. Selon le contexte, une donnée manquante peut également être représentée par None ou pd.NA.

On peut ensuite utiliser `.sum()` pour compter les valeurs manquantes dans chaque colonne.

````python
# Nombre de valeurs manquantes, par colonne

df.isna().sum()
````

**Comment lire ce résultat ?** `.isna()` donne un tableau de `True`/`False` de la même taille que `df`. En lui appliquant `.sum()`, on additionne ces `True`/`False` colonne par colonne et comme `True` vaut `1` et `False` vaut `0` en Python, cette somme donne directement le **nombre** de valeurs manquantes par colonne. C'est une astuce de Pandas qu'il faut bien comprendre : `.sum()` sur des booléens = un comptage.

On voit ici que `satisfaction_rh` a quelques valeurs manquantes (8). C'est cohérent avec ce que `.info()` nous montrait déjà à la section 1.

### 3.2. Que faire des valeurs manquantes ?

Après avoir repéré les valeurs manquantes, il faut décider comment les traiter. Il n'existe pas de règle universelle ; le choix dépend de **la nature des données, de la quantité de valeurs manquantes et de l'objectif de l'analyse**.

Deux approches sont généralement utilisées.

#### 3.2.1. Option 1 : supprimer les observations concernées

La méthode `.dropna()` permet de supprimer les lignes contenant des valeurs manquantes. Avec `subset`, on peut préciser la colonne à prendre en compte :

````python
# Option 1 : supprimer les lignes qui ont une valeur manquante dans une colonne donnée

df_sans_na = df.dropna(subset=["satisfaction_rh"])
print(f"{len(df)} lignes avant, {len(df_sans_na)} lignes après suppression")
````

Ici, seules les lignes pour lesquelles satisfaction_rh est manquante sont supprimées.

> Cependant, supprimer des observations signifie aussi perdre de l'information. Si les valeurs manquantes sont liées aux caractéristiques des répondants ou des entreprises, cette suppression peut introduire un biais dans les résultats.

#### 3.2.2. Option 2 : remplacer les valeurs manquantes

Une autre possibilité consiste à remplacer les valeurs manquantes par une valeur estimée ou calculée. Par exemple, on peut remplacer les valeurs manquantes par la moyenne de la colonne :

````python
# Option 2 : remplacer les valeurs manquantes par une valeur (ex. la moyenne de la colonne)

moyenne_satisfaction = df["satisfaction_rh"].mean()

df_rempli = df.copy()  # on travaille sur une copie pour ne pas modifier df directement

df_rempli["satisfaction_rh"] = (
    df_rempli["satisfaction_rh"].fillna(moyenne_satisfaction)  # fillna() remplace ici chaque valeur manquante par la moyenne calculée.
)

print(f"Valeurs manquantes restantes : {df_rempli['satisfaction_rh'].isna().sum()}")  # Pour vérifier qu'il ne reste plus de valeurs manquantes :
````

**Comment choisir entre les deux ?**

- **Supprimer** (`dropna`) est simple et sûr si vous avez très peu de valeurs manquantes et beaucoup de données, mais vous perdez de l'information. Et si les valeurs manquantes ne sont pas "au hasard" (ex. les entreprises les moins satisfaites répondent moins souvent), vous introduisez un biais.
- **Remplacer** (`fillna`) conserve toutes vos lignes, mais vous "inventez" une valeur. À utiliser avec prudence, en étant transparent sur ce choix méthodologique dans votre recherche.

> [!NOTE]
> Traiter les valeurs manquantes est un choix méthodologique, pas seulement une opération technique. Le chercheur doit comprendre pourquoi des données sont manquantes avant de décider de les supprimer ou de les remplacer.

*Pourquoi `.copy()` à la ligne 3 ?* Sans `.copy()`, `df_rempli` ne serait qu'une autre façon de désigner les mêmes données que `df` en mémoire (ce qu'on appelle une "vue") ; et les modifier via `df_rempli` **modifierait aussi `df`** sans que vous vous en rendiez compte. `.copy()` crée une véritable copie indépendante.

### 3.3. Repérer et traiter les doublons

Un doublon est **une observation qui apparaît plusieurs fois** dans un jeu de données. Dans Pandas, la méthode `.duplicated()` permet de repérer les lignes qui sont des copies exactes d'une ligne déjà présente.

#### 3.3.1. Compter les doublons

Pour connaître le nombre de lignes en double :


````python
# Nombre de lignes strictement identiques à une autre ligne

df.duplicated().sum()  # df.duplicated() examine chaque ligne et retourne True lorsqu'elle est identique à une ligne précédente. Et avec .sum(), on compte les True.
````

Cela signifie ici que **4 lignes** sont des doublons d'une ligne précédente.

> [!IMPORTANT]
> `.duplicated()` considère ici qu'une ligne est un doublon lorsque toutes ses colonnes sont identiques à celles d'une ligne précédente.

#### 3.3.2. Voir les doublons

Pour afficher les lignes identifiées comme doublons :

````python
# Voir les doublons

df[df.duplicated()]
````

Cette instruction utilise le résultat de `df.duplicated()` comme filtre pour ne conserver que les lignes pour lesquelles le résultat est `True`.

#### 3.3.3. Supprimer les doublons

La méthode `.drop_duplicates()` permet de supprimer les lignes en double :

````python
# Supprimer les doublons

df_sans_doublons = df.drop_duplicates()  # Par défaut, Pandas conserve la première occurrence et supprime les suivantes.

print(f"{len(df)} lignes avant, {len(df_sans_doublons)} lignes après suppression des doublons") # Vérifier le nombre de lignes avant et après
````

Cela signifie ici que 4 lignes en double ont été supprimées.

> **Attention aux faux doublons**. Deux observations ne sont pas nécessairement des doublons simplement parce qu'elles ont certaines valeurs identiques. Il est donc important de comprendre ce qui définit une observation unique dans vos données avant de supprimer des doublons.

### 3.4. Vérifier les types

Avant d'analyser une colonne, il est important de vérifier que son type correspond bien à la nature des données qu'elle contient. Par exemple, une variable représentant le nombre d'employés devrait être numérique :

```
effectifs
25
42
18
63
```

Mais il peut arriver qu'une colonne soit enregistrée comme du texte (`str`) à cause du fichier source ou de certaines valeurs inattendues. Par exemple :

```
effectifs
25
42
"18 "
63
```

L'espace après 18 peut causer des problèmes lors de l'importation ou du traitement des données. Dans ces cas, `.describe()` ne fonctionnera pas comme attendu sur cette colonne, et les calculs (additions, moyennes...) échoueront ou donneront des résultats absurdes.

Pour cela, il faut toujours faire une vérification avec la méthode `.info()` (vu en section 1) pour connaître le type de chaque colonne. Ensuite, lorsque nous savons qu'une colonne doit être numérique, nous pouvons utiliser la méthode `pd.to_numeric()`.

Exemple illustratif : Notre colonne est déjà propre. Le code suivant ne fera donc rien de visible ici. Cependant, c'est le réflexe à avoir face à un vrai fichier où ce ne serait pas le cas.


````python
df["effectifs"] = pd.to_numeric(
    df["effectifs"], errors="coerce"
)
````

Cette instruction demande à Pandas de convertir les valeurs de `effectifs` en nombres.

**Que signifie `errors="coerce"` ?**

Certaines valeurs peuvent être impossibles à convertir. Si on a par exemple :

```
25
42
NC
63
```

Avec l'instruction :

```python
pd.to_numeric(df["effectifs"], errors="coerce")
```

Pandas va convertir les valeurs numériques normalement et transforme les valeurs impossibles à convertir en NaN :

```
25
42
NaN
63
```

Ainsi, le programme ne s'arrête pas à cause d'une valeur problématique.

> [!IMPORTANT]
> Attention, `errors="coerce"` ne corrige pas la donnée. Il transforme simplement les valeurs impossibles à convertir en valeurs manquantes. Il faut donc vérifier ensuite combien de NaN ont été créés, avec : `df["effectifs"].isna().sum()`. Si le résultat est, par exemple, `1`, cela signifie qu'une valeur n'a pas pu être convertie.

***Retenez bien, qu'avant de calculer une moyenne, une somme ou de réaliser une analyse statistique, il faut s'assurer que les variables ont le bon type.***

## 4. Créer et transformer des variables

En analyse de données, il est fréquent de devoir **créer de nouvelles variables ou transformer celles qui existent déjà**. Par exemple, à partir du chiffre d'affaires et du nombre de salariés, nous pouvons calculer un indicateur de productivité. Nous pouvons également transformer une variable numérique en catégorie ou regrouper plusieurs modalités d'une variable. Avec Pandas, ces opérations peuvent généralement être réalisées directement sur les colonnes d'un DataFrame.

### 4.1. Créer une colonne calculée

Pour créer une nouvelle colonne, on l'assigne comme s'il s'agissait d'une variable Python classique ; sauf qu'ici, l'opération s'applique **automatiquement à toutes les lignes en une fois** (on parle d'opération "vectorisée") :


````python

df["ca_par_salarie"] = df["chiffre_affaires_keur"] / df["effectifs"]          # Ici, nous créons la colonne ca_par_salarie en divisant le chiffre d'affaires par le nombre de salariés.

# On peut ensuite vérifier le résultat
df[["entreprise_id", "annee", "chiffre_affaires_keur", "effectifs", "ca_par_salarie"]].head()
````

**Qu'est-ce qu'une opération vectorisée ?**

Avec l'instruction `df["chiffre_affaires_keur"] / df["effectifs"]`, Pandas effectue automatiquement la division pour chaque ligne correspondante des deux colonnes. Si le DataFrame contient 404 lignes, Pandas effectue donc le calcul pour les 404 lignes sans que nous ayons besoin d'écrire une boucle `for`. On parle d'opération vectorisée.

Avec une boucle, il faudrait parcourir les lignes une par une. Avec Pandas, on peut généralement effectuer directement l'opération sur les colonnes :

```python
df["colonne_a"] / df["colonne_b"]
```

C'est l'un des grands avantages de Pandas. Le code est alors plus court, plus lisible et plus efficace que lorsqu'on traite manuellement chaque ligne.

### 4.2. Utiliser  `apply()` pour des transformations personnalisées

Toutes les transformations ne se résument pas à une addition, une soustraction ou une division. Supposons que nous souhaitions classer les entreprises selon leur nombre de salariés :

Pour des transformations plus complexes qu'une simple opération mathématique, `apply()` permet d'exécuter une fonction personnalisée sur chaque ligne (ou chaque valeur) d'une colonne.

- moins de 50 → Petite ;
- de 50 à 249 → Moyenne ;
- 250 ou plus → Grande.

Nous pouvons d'abord créer une fonction :


````python
def categoriser_taille(effectif):
    if effectif < 50:
        return "Petite"
    elif effectif < 250:
        return "Moyenne"
    else:
        return "Grande"

df["taille_entreprise"] = df["effectifs"].apply(categoriser_taille)
df[["entreprise_id", "effectifs", "taille_entreprise"]].head()
````

Cette fonction est celle que nous avons apprise au module 01 (sections fonctions). Elle reçoit une valeur puis retourne un résultat. Nous pouvons ensuite l'appliquer à toute la colonne avec `apply()` :

`df["taille_entreprise"] = df["effectifs"].apply(categoriser_taille)`

Pandas prend chaque valeur de `effectifs`, la transmet à `categoriser_taille()` et place le résultat dans la nouvelle colonne. On peut vérifier le résultat avec :

```python
df[
    ["entreprise_id", "effectifs", "taille_entreprise"]
].head()
```

Comme : « Applique la fonction categoriser_taille à chaque valeur de la colonne `effectifs`. »

`apply()` est particulièrement utile lorsque la transformation nécessite une logique personnalisée qui ne peut pas être exprimée simplement avec une opération vectorisée.


### 4.3. Recoder des catégories avec `map()`

En recherche, il est également fréquent de devoir recoder une variable catégorielle. Par exemple, nous pouvons souhaiter regrouper plusieurs régions en deux grandes zones pour simplifier l'analyse. Dans ce cas, on peut écrire le code suivant :

````python

zones = {
    "Île-de-France": "Zone A",
    "Auvergne-Rhône-Alpes": "Zone A",
    "Nouvelle-Aquitaine": "Zone B",
    "Hauts-de-France": "Zone B",
    "Occitanie": "Zone B",
}

df["zone"] = df["region"].map(zones)

df[["region", "zone"]].drop_duplicates() # Vérifier les correspondances
````

Ici, `zones` est un dictionnaire Python. Chaque région constitue une clé et la zone correspondante constitue sa valeur. Nous pouvons appliquer cette correspondance à la colonne region avec `map()`. Ainsi, `map()` prend un dictionnaire et remplace chaque valeur de la colonne par sa correspondance dans le dictionnaire ; exactement la logique des dictionnaires vue au module 01, appliquée ici à une colonne entière.

Cette opération permet notamment de **recoder des modalités, regrouper des catégories ou simplifier une variable pour une analyse**.

### 4.4. `apply()` ou `map()` ?

Retenez simplement ceci :

- opération mathématique simple → opération vectorisée ;
- transformation nécessitant une logique personnalisée → apply() ;
- correspondance entre des catégories et de nouvelles valeurs → map().

Il existe souvent plusieurs façons d'effectuer une même transformation dans Pandas. L'objectif n'est pas de mémoriser toutes les méthodes, mais de comprendre quelle approche correspond au problème que l'on cherche à résoudre.

## 5. Grouper et agréger avec `groupby`

En recherche, on cherche généralement à comparer des groupes d'observations. Par exemple :

- les entreprises qui pratiquent le télétravail et celles qui ne le pratiquent pas ;
- les entreprises de différents secteurs ;
- les entreprises selon leur taille ;
- les répondants selon leur catégorie d'âge.

Pandas permet de réaliser ce type de comparaison avec la méthode `groupby()`.

C'est ici que notre question de recherche commence vraiment à prendre forme : **`groupby` est l'outil qui permet de comparer des groupes entre eux** ; exactement ce dont on a besoin pour comparer les entreprises en télétravail et les autres.

> [!NOTE]
> `groupby()` permet de regrouper les observations selon une ou plusieurs variables, puis d'effectuer un calcul séparément pour chaque groupe.

### 5.1. Le principe : "diviser, appliquer, combiner"

`groupby()` fonctionne toujours selon le même principe en trois étapes, même si on l'écrit en une seule ligne de code :

1. **Diviser** les données en groupes (ex. un groupe par valeur de `teletravail`)
2. **Appliquer** un calcul à chaque groupe séparément (ex. la moyenne)
3. **Combiner** les résultats dans un nouveau tableau, un résultat par groupe

Imaginons que notre DataFrame contienne une variable teletravail indiquant si une entreprise pratique ou non le télétravail. Nous pouvons calculer le chiffre d'affaires moyen selon le télétravail :

````python

df.groupby("teletravail")["chiffre_affaires_keur"].mean()
````

**Décomposons cette instruction**

Pandas :

- sépare `df` en deux groupes (les lignes où `teletravail` vaut 0, celles où il vaut 1) ;
- calcule la moyenne de `chiffre_affaires_keur` séparément dans chacun des deux groupes ;
- puis affiche les deux résultats côte à côte, avec `.mean()`.

Nous obtenons donc le chiffre d'affaires moyen pour les entreprises sans télétravail (0) et avec télétravail (1).

Cela permet déjà de faire une première comparaison descriptive. Mais attention ! Cette comparaison ne permet pas à elle seule de conclure que le télétravail contribue à une augmentation du chiffre d'affaires. Une différence de moyenne peut avoir de nombreuses autres explications. Les tests statistiques et les modèles étudiés plus loin permettront de vérifier cela.

### 5.2. Grouper selon plusieurs variables

Il est possible de créer des groupes à partir de plusieurs variables. Par exemple, nous pouvons comparer la rentabilité selon le secteur et la pratique du télétravail :


````python
df.groupby(["secteur", "teletravail"])["rentabilite"].mean()
````

Pandas crée alors un groupe pour chaque combinaison de `secteur` et `teletravail` ; chaque ligne correspondant à une combinaison de catégories.

### 5.3. Calculer plusieurs statistiques avec  `.agg() `

Nous ne sommes pas obligés de calculer uniquement une moyenne. La méthode `.agg()` permet d'appliquer plusieurs fonctions d'agrégation à une variable :


````python
df.groupby(["secteur", "teletravail"])["rentabilite"].agg(["mean", "std", "count"])
````

Nous obtenons notamment :

- `mean` → la moyenne ;
- `std` → l'écart-type, qui mesure la dispersion des valeurs autour de la moyenne ;
- `count` → le nombre d'observations du groupe.

Le nombre d'observations est particulièrement important. Par exemple, une moyenne calculée sur :

```
Groupe A → 150 observations
Groupe B → 4 observations
```

ne doit pas être interprétée de la même manière.


### 5.4. Présenter les résultats avec `pivot_table()`

La méthode `groupby()` produit parfois un résultat difficile à lire lorsque plusieurs variables sont utilisées. Dans ces cas, Pandas propose la méthode `pivot_table()` pour présenter certaines analyses sous la forme d'un tableau croisé. Par exemple :


````python
pd.pivot_table(
    df,
    values="satisfaction_rh",
    index="secteur",                  # les lignes du tableau croisé
    columns="teletravail",        # les colonnes du tableau croisé
    aggfunc="mean",
)
````

Dans le code :

- `values="satisfaction_rh"` → la variable que nous voulons analyser ;
- `index="secteur"` → les secteurs apparaissent en lignes ;
- `columns="teletravail"` → les catégories de télétravail apparaissent en colonnes ;
- `aggfunc="mean"` → nous calculons la moyenne.

On obtient alors un vrai tableau croisé secteur × télétravail, avec la satisfaction RH moyenne dans chaque case. On peut ensuite comparer rapidement la satisfaction moyenne selon le secteur et la pratique du télétravail.

## 6. Croiser plusieurs jeux de données

En recherche, les données dont nous avons besoin sont rarement réunies dans un seul fichier. Par exemple, nous pouvons disposer de :

- données financières dans un fichier ;
- données RH dans un autre ;
- informations géographiques dans un troisième ;
- données collectées à plusieurs années dans plusieurs fichiers.

Il faut alors pouvoir réunir ces données. Pandas propose principalement deux outils pour cela :

- `merge()` qui permet de croiser des données à partir d'une variable commune ;
- `concat()` qui permet d'assembler plusieurs DataFrames en les empilant.

### 6.1. Croiser des données avec `merge()`

Pour illustrer le principe, imaginons que nous disposions de deux DataFrames. Le premier contient nos données principales ; par exemple :

```
entreprise_id    chiffre_affaires
E001             1500
E002             2300
E003             1800
E004             2450
```

Le second contient des informations sur le siège social ; par exemple :

```
entreprise_id    ville_siege
E001             Toulouse
E002             Lyon
E003             Bordeaux
E004             Lille
```

Les deux tableaux ont une colonne commune : `entreprise_id`. Cette colonne permet d'identifier les entreprises présentes dans les deux jeux de données. Nous pouvons créer le second DataFrame ainsi :

````python
sieges_sociaux = pd.DataFrame({
    "entreprise_id": ["E001", "E002", "E003", "E004"],
    "ville_siege": ["Toulouse", "Lyon", "Bordeaux", "Lille"],
})

sieges_sociaux
````

Pour croiser les deux DataFrames, nous allons utiliser `merge()` :

`merge()` croise deux DataFrames sur une colonne commune (ici, `entreprise_id`), un peu comme une jointure SQL si vous connaissez déjà ce vocabulaire ; sinon, ne vous en inquiétez pas, l'important est de comprendre le principe ci-dessous.


````python
df_avec_siege = df.merge(
    sieges_sociaux,
    on="entreprise_id",                                    # Le paramètre on indique la colonne qui permet de faire correspondre les deux DataFrames.
    how="left"
)

df_avec_siege[["entreprise_id", "annee", "secteur", "ville_siege"]].head(8)  # Pour vérifier le résultat
````

**Le paramètre `how`**

Le paramètre how indique quelles lignes doivent être conservées lorsqu'une correspondance existe ou non entre les deux DataFrames. Les quatre principales options sont :

| `how=` | Comportement |
|---|---|
| `"left"` | Garde **toutes** les lignes du DataFrame de gauche (`df`), même celles sans correspondance dans `sieges_sociaux` (les colonnes ajoutées seront alors `NaN`) |
| `"right"` | Symétrique : garde toutes les lignes du DataFrame de droite |
| `"inner"` | Garde **seulement** les lignes présentes des deux côtés |
| `"outer"` | Garde **toutes** les lignes des deux côtés, `NaN` là où il n'y a pas de correspondance |

Dans notre exemple, `sieges_sociaux` ne contient que 4 entreprises sur les 80 du jeu de données complet. Avec `how="left"`, on garde donc bien toutes les lignes de `df`. Si une entreprise n'existe pas dans `sieges_sociaux`, Pandas ne peut pas trouver de ville et indique une valeur manquante : `NaN`.

On peut ensuite vérifier combien de correspondances n'ont pas été trouvées :


````python

df_avec_siege["ville_siege"].isna().sum()
````

### 6.2. Pourquoi vérifier le résultat d'un `merge()` ?

Un `merge()` peut modifier le nombre de lignes du DataFrame. Il faudra donc toujours vérifier le résultat d'une jointure.


````python

print("Avant :", len(df))
print("Après :", len(df_avec_siege))
````

Si le nombre de lignes a changé alors que vous ne vous y attendiez pas, il faut chercher pourquoi. Cela peut être lié à des identifiants absents ou à des valeurs répétées dans la colonne utilisée pour la jointure.


### 6.3. Empiler des données avec `concat()`

`merge()` permet de croiser des colonnes provenant de différents jeux de données. Mais que faire lorsque nous avons plusieurs DataFrames qui ont les mêmes colonnes et que nous souhaitons simplement les réunir ? C'est le rôle de `concat()`.

Imaginons que nous ayons un DataFrame pour 2019 et un autre pour 2020. Nous pouvons les empiler avec ce code :


````python
df_2019 = df[df["annee"] == 2019]
df_2020 = df[df["annee"] == 2020]

df_2019_2020 = pd.concat([df_2019, df_2020])    # Les lignes du deuxième DataFrame sont ajoutées à la suite de celles du premier.

print(f"{len(df_2019)} + {len(df_2020)} = {len(df_2019_2020)} lignes après concat")   # Vérifier le résultat
````

### 6.4. `merge()` ou `concat()` ?

Une distinction simple permet de retenir l'essentiel :

`merge()` → **je veux croiser des informations**

```
Base financière        Base RH
      ↓                    ↓
      └──── entreprise_id ─┘
               ↓
        Base enrichie
```

`concat()` → **je veux empiler des données**

```
Données 2019
     ↓
Données 2020
     ↓
Données 2019 + 2020
```

> [!NOTE]
> Retene donc qu'il faut utiliser `merge()` lorsque plusieurs jeux de données doivent être reliés grâce à une clé commune. Utilisez `concat()` lorsque vous souhaitez principalement réunir des DataFrames ayant une structure similaire. En recherche, cette distinction est importante, car la manière dont les jeux de données sont assemblés peut avoir un impact sur l'échantillon analysé.


## 7. Gérer les dates

Les dates sont très fréquentes dans les données de recherche : date d'une enquête, date de création d'une entreprise, date d'une transaction, année d'observation, date de publication d'un document, etc. Pour pouvoir manipuler correctement des dates dans Pandas, il faut que la colonne soit reconnue comme une date et non comme du simple texte.

Notre jeu de données contient une variable `annee`, qui indique l'année d'observation :


````python
df["annee"].head()
````

Cette variable est actuellement un entier (`int64`), et non une date. Cela convient pour certaines analyses, mais certaines opérations nécessitent de travailler avec de véritables dates :

- extraire un mois ou un jour ;
- calculer une durée entre deux dates ;
- comparer des dates, etc.

Pour illustrer ces opérations, nous allons créer une date à partir de `annee`.

### 7.1. Créer une date à partir d'une année

Imaginons que nous souhaitions représenter le début de chaque année d'observation. Nous pouvons créer une chaîne de caractères contenant le 1er janvier de chaque année :

````python
dates_texte = df["annee"].astype(str) + "-01-01"

dates_texte.head()
````

Pour l'instant, ces valeurs sont encore du texte. Nous allons donc les convertir en véritables dates avec `pd.to_datetime()`.

### 7.2. Convertir en `datetime`

`pd.to_datetime()` convertit une colonne de texte en un vrai type "date", ce qui débloque tout un ensemble d'opérations :


````python

df["date_observation"] = pd.to_datetime(dates_texte)

df[["annee", "date_observation"]].head()    # Vérifier le résultat
````

````python
df[["date_observation"]].info()  # Vérifier le type de colonne
````

On obtient `datetime64[us]`. Cela signifie que Pandas reconnaît désormais `date_observation` comme une date, et non comme du texte.

### 7.3. Extraire des composantes d'une date

Une fois la colonne au format `datetime`, nous pouvons utiliser `.dt` pour accéder à différentes composantes de la date. Par exemple : 

````python
df["annee_observation"] = df["date_observation"].dt.year     # Pour récupérer l'année

df["mois_observation"] = df["date_observation"].dt.month   # Pour récupérer le mois

df["jour_observation"] = df["date_observation"].dt.day          # Pour récupérer le jour

# Afficher les résultats

df[["date_observation", "annee_observation", "mois_observation", "jour_observation"]].head()
````

### 7.4. Calculer une durée

L'un des intérêts importants des dates est de pouvoir calculer des écarts de temps. Par exemple, nous pouvons calculer combien d'années séparent l'année de création de l'entreprise de l'année d'observation :

````python

df["age_entreprise"] = (
    df["annee"] - df["annee_creation"]
)
````

Ici, nous n'avons même pas besoin de `datetime`, car nous travaillons simplement avec deux années. Par exemple :

```
annee_creation    annee    age_entreprise
2018              2019     1
2018              2020     2
2018              2021     3
```

Cette nouvelle variable peut être particulièrement intéressante pour notre analyse : l'âge de l'entreprise pourrait constituer un facteur à prendre en compte lorsque nous comparons les entreprises.

### 7.5. Trouver la période couverte par les données

Nous pouvons également utiliser `.min()` et `.max()` sur `annee` pour connaître la période couverte par notre jeu de données :

````python
df["annee"].min()
````

````python
df["annee"].max()
````

Cela permet de déterminer rapidement la première et la dernière année disponibles. Nous pouvons faire la même chose avec notre colonne `date_observation` :


````python
df["date_observation"].min()
````

````python
df["date_observation"].max()
````
