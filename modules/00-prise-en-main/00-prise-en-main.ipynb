## 2. Se repérer dans Jupyter et JupyterLab

### Jupyter, JupyterLab, notebook : un peu de vocabulaire

- **Un notebook** (fichier `.ipynb`) est un document qui mélange code exécutable, texte explicatif et visualisations dans le navigateur. C'est l'outil que l'on utilise généralement pour apprednre Python, explorer des données et pour faire de la data science. C'est le format choisi pour tous les modules de ce cours.
- **Jupyter** est un outil informatique ouvert qui sert à créer des documents interactifs appelés *notebooks*.
- **JupyterLab** est l'interface (l'application) dans laquelle vous pouvez ouvrir et exécuter des notebooks. C'est ce que vous voyez à l'écran quand vous lancez `jupyter lab`. Il en existe une version plus ancienne et plus simple, *Jupyter Notebook* (ou *Notebook classique*). Les deux ouvrent les mêmes fichiers `.ipynb`. Cependant, JupyterLab est plus complet (plusieurs onglets, explorateur de fichiers intégré, terminal, etc.).
- **Google Colab**, ou Colaboratory, est un service gratuit en ligne de Google qui permet d'écrire et d'exécuter du code Python dans un navigateur web. Il fonctionne sans installation locale et offre un accès à des puces puissantes comme les GPU et TPU pour l'intelligence artificielle. On peut la considérer comme une version hébergée par Google, modifiée et prête à l'emploi de l'écosystème Jupyter. Avec cette solution, vous n'avez rien à installer sur votre ordinateur ; tout se passe directement dans votre navigateur.

Dans ce cours, "Jupyter" et "JupyterLab" sont utilisés à peu près indifféremment pour désigner l'endroit où vous travaillez.

### Tour rapide de l'interface JupyterLab

Si vous êtes en local avec JupyterLab, voici ce que vous devriez voir :

- **À gauche** : un explorateur de fichiers, qui montre le dossier dans lequel vous avez lancé `jupyter lab` (normalement, la racine de ce repo). C'est là que vous naviguez entre les modules.
- **En haut** : une barre de menus (File, Edit, View, Run, Kernel...) et, pour chaque notebook ouvert, une barre d'outils avec des boutons (sauvegarder, ajouter une cellule, exécuter, arrêter, redémarrer...).
- **Au centre** : la zone principale, avec un onglet par fichier ouvert. Vous pouvez avoir plusieurs notebooks ouverts en même temps, chacun dans son onglet.
- **En bas à droite de chaque notebook** : le nom du kernel utilisé (normalement `Python 3`) et un petit cercle indiquant s'il est occupé (rempli) ou disponible (vide).

### Les types de cellules

Un notebook est composé de **cellules**, empilées les unes sous les autres. Il en existe principalement deux types, que vous utiliserez pour ce cours :

- des cellules de **texte** (Markdown), comme celle-ci, pour les explications, les titres, la mise en forme ;
- des cellules de **code**, comme celles qui suivent, pour exécuter du code Python.

*(Il existe un troisième type, les cellules "Raw", pour du texte non interprété. Vous ne devriez pas en avoir besoin dans le cadre de ce cours.)*

Vous reconnaissez le type d'une cellule à sa mise en forme : le texte Markdown s'affiche formaté (titres en gras, listes...), tandis qu'une cellule de code garde une police à chasse fixe et un fond légèrement différent, avec `[ ]` ou `[n]` sur sa gauche une fois exécutée.

### Mode édition vs mode commande

Une cellule sélectionnée peut être dans l'un de ces deux **modes**, reconnaissables à la couleur de son cadre :

- **Mode édition** (cadre généralement vert, curseur clignotant visible) : vous pouvez taper du texte ou du code à l'intérieur de la cellule. On y entre en cliquant *dans* le contenu d'une cellule ou en appuyant sur `Entrée` quand elle est sélectionnée.
- **Mode commande** (cadre généralement bleu, pas de curseur) : la cellule est sélectionnée, mais vous ne pouvez rien saisir à l'intéreieur. Les touches du clavier deviennent alors des raccourcis qui agissent sur la cellule elle-même (la supprimer, en ajouter une nouvelle, etc.). On y entre en appuyant sur `Échap` ou en cliquant sur la marge à gauche de la cellule (en dehors du texte).

### Quelques raccourcis utiles (en mode commande, donc après `Échap`)

| Raccourci | Action |
|---|---|
| `Shift + Entrée` | Exécuter la cellule et passer à la suivante |
| `Ctrl + Entrée` (`Cmd + Entrée` sur Mac) | Exécuter la cellule et rester dessus |
| `a` | Insérer une nouvelle cellule **au-dessus** (*above*) |
| `b` | Insérer une nouvelle cellule **en dessous** (*below*) |
| `d`, `d` (deux fois) | Supprimer la cellule sélectionnée |
| `m` | Transformer la cellule en cellule Markdown |
| `y` | Transformer la cellule en cellule de code |

Pas besoin de les mémoriser pour le moment. Vous les retiendrez naturellement avec la pratique. Vous pouvez toujours faire la même chose via les boutons de la barre d'outils ou le menu, si vous préférez.

### Le kernel (noyau)

Le **kernel** est le processus qui exécute réellement votre code Python en arrière-plan. C'est lui qui garde en mémoire vos variables entre deux cellules. Chaque notebook ouvert a son propre kernel, indépendant des autres.

Il faut garder en tête que le kernel n'a **aucune idée de l'ordre visuel** des cellules dans le notebook. Il exécute les cellules dans l'ordre où *vous* les lancez, pas dans l'ordre où elles apparaissent à l'écran.

### L'ordre d'exécution compte

Le nombre entre crochets à gauche d'une cellule de code (ex. `[1]`) indique l'**ordre dans lequel les cellules ont été exécutées**. Il ne s'agit donc pas de la position dans le notebook. Si vous exécutez les cellules dans le désordre, l'état de votre programme peut ne plus correspondre à ce qui est affiché.

**Bonne pratique** : de temps en temps (et systématiquement avant de partager un notebook), redémarrez le noyau et exécutez toutes les cellules dans l'ordre, via le menu *Kernel/Noyau → Restart & Run All*. Cela garantit que votre notebook fonctionne bien de haut en bas, sans dépendre d'exécutions précédentes.

Essayons un exemple :


````python
# Exécutez d'abord cette cellule...
valeur = 10
````

### Redémarrer le noyau

Si votre notebook se comporte bizarrement (variables qui ne correspondent plus à ce que vous attendez, erreurs étranges), le réflexe à avoir est de **redémarrer le noyau** (*Kernel → Restart*) puis de tout ré-exécuter depuis le début. C'est souvent la solution la plus rapide, même si elle semble radicale.
