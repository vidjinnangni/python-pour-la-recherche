## 2. Se repérer dans Jupyter et JupyterLab

### 2.1. Jupyter, JupyterLab, notebook : un peu de vocabulaire

- **Un notebook** (fichier `.ipynb`) est un document qui mélange code exécutable, texte explicatif et visualisations dans le navigateur. C'est l'outil que l'on utilise généralement pour apprendre Python, explorer des données et pour faire de la data science. C'est le format choisi pour tous les modules de ce cours.
- **Jupyter** est un ensemble d'outils permettant de créer et d'exécuter des notebooks interactifs. Le nom Jupyter vient historiquement de la combinaison de trois langages : Julia, Python et R. Aujourd'hui, Jupyter est utilisé avec de nombreux langages, mais Python est l'un de ses usages les plus courants.
- **JupyterLab** est l'interface (l'application) dans laquelle vous pouvez ouvrir et exécuter des notebooks. C'est ce que vous voyez à l'écran quand vous lancez `jupyter lab`. Il en existe une version plus ancienne et plus simple, *Jupyter Notebook* (ou *Notebook classique*). Les deux ouvrent les mêmes fichiers `.ipynb`. Cependant, JupyterLab est plus complet (plusieurs onglets, explorateur de fichiers intégré, terminal, etc.).
- **Google Colab**, ou Colaboratory, est un service gratuit en ligne de Google qui permet d'écrire et d'exécuter du code Python dans un navigateur web. Il fonctionne sans installation locale et offre un accès à des puces puissantes comme les GPU et TPU pour l'intelligence artificielle. On peut la considérer comme une version hébergée par Google, modifiée et prête à l'emploi de l'écosystème Jupyter. Avec cette solution, vous n'avez rien à installer sur votre ordinateur ; tout se passe directement dans votre navigateur.

Dans ce cours, "Jupyter" et "JupyterLab" sont utilisés à peu près indifféremment pour désigner l'endroit où vous travaillez.

### 2.2. Tour rapide de l'interface JupyterLab

Si vous êtes en local avec JupyterLab, voici ce que vous devriez voir :

- **À gauche** : un explorateur de fichiers, qui montre le dossier dans lequel vous avez lancé `jupyter lab` (normalement, la racine de ce repo). C'est à partir de là que vous pourrez naviguer entre les modules.
- **En haut** : une barre de menus (File, Edit, View, Run, Kernel...) et, pour chaque notebook ouvert, une barre d'outils avec des boutons (sauvegarder, ajouter une cellule, exécuter, arrêter, redémarrer...).
- **Au centre** : la zone principale, avec un onglet par fichier ouvert. Vous pouvez avoir plusieurs notebooks ouverts en même temps, chacun dans son onglet.
- **En bas à droite de chaque notebook** : le nom du kernel utilisé (normalement `Python 3`) et un petit cercle indiquant s'il est occupé (rempli) ou disponible (vide). **À retenir :** le notebook est le document dans lequel vous écrivez votre code ; le kernel est ce qui exécute ce code.

### 2.3. Les types de cellules

Un notebook est composé de **cellules**, empilées les unes sous les autres. Il en existe principalement deux types, que vous utiliserez pour ce cours :

- des cellules de **texte** (Markdown), comme celle-ci, pour les explications, les titres, la mise en forme ;
- des cellules de **code**, comme celles qui suivent, pour exécuter du code Python.

*(Il existe un troisième type, les cellules "Raw", pour du texte non interprété. Vous ne devriez pas en avoir besoin dans le cadre de ce cours.)*

Vous reconnaissez le type d'une cellule à sa mise en forme : le texte Markdown s'affiche formaté (titres en gras, listes...), tandis qu'une cellule de code garde une police à chasse fixe et un fond légèrement différent, avec `[ ]` ou `[n]` sur sa gauche une fois exécutée.

### 2.4. Mode édition vs mode commande

Une cellule sélectionnée peut être dans l'un de ces deux **modes**, reconnaissables à la couleur de son cadre :

- **Mode édition** (cadre généralement vert, curseur clignotant visible) : vous pouvez taper du texte ou du code à l'intérieur de la cellule. On y entre en cliquant *dans* le contenu d'une cellule ou en appuyant sur `Entrée` quand elle est sélectionnée.
- **Mode commande** (cadre généralement bleu, pas de curseur) : la cellule est sélectionnée, mais vous ne pouvez rien saisir à l'intéreieur. Les touches du clavier deviennent alors des raccourcis qui agissent sur la cellule elle-même (la supprimer, en ajouter une nouvelle, etc.). On y entre en appuyant sur `Échap` ou en cliquant sur la marge à gauche de la cellule (en dehors du texte).

### 2.5. Quelques raccourcis utiles (en mode commande, donc après `Échap`)

| Raccourci | Action |
|---|---|
| `Shift + Entrée` | Exécuter la cellule et passer à la suivante |
| `Ctrl + Entrée` (`Cmd + Entrée` sur Mac) | Exécuter la cellule et rester dessus |
| `a` | Insérer une nouvelle cellule **au-dessus** (*above*) |
| `b` | Insérer une nouvelle cellule **en dessous** (*below*) |
| `d`, `d` (deux fois) | Supprimer la cellule sélectionnée |
| `m` | Transformer la cellule en cellule Markdown |
| `y` | Transformer la cellule en cellule de code |

Pas besoin de les mémoriser pour le moment. Vous les retiendrez naturellement avec la pratique. Vous pouvez toujours faire la même chose via les boutons de la barre d'outils ou le menu, si vous préférez.

### 2.6. Le kernel (noyau)

Le **kernel** est le processus qui exécute réellement votre code Python en arrière-plan. C'est lui qui garde en mémoire vos variables entre deux cellules. Chaque notebook ouvert a son propre kernel, indépendant des autres. Il faut garder en tête que le kernel n'a **aucune idée de l'ordre visuel** des cellules dans le notebook. Il exécute les cellules dans l'ordre dans lequel *vous* les lancez, pas dans l'ordre dans lequel elles apparaissent à l'écran.

- `Ctrl + Entrée` (`Cmd + Entrée` sur Mac) : exécute la cellule et reste dessus

### 2.8. L'ordre d'exécution

Le nombre entre crochets à gauche d'une cellule de code (ex. `[1]`) indique l'**ordre dans lequel les cellules ont été exécutées**. Il ne s'agit donc pas de la position dans le notebook. Si vous exécutez les cellules dans le désordre, l'état de votre programme peut ne plus correspondre à ce qui est affiché.

**Bonne pratique** : de temps en temps (et systématiquement avant de partager un notebook), redémarrez le noyau et exécutez toutes les cellules dans l'ordre, via le menu *Kernel/Noyau → Restart & Run All Cells...*. Cela garantit que votre notebook fonctionne bien de haut en bas, sans dépendre d'exécutions précédentes.

Essayons un exemple :


````python
# Exécutez d'abord cette cellule...
valeur = 10
````

### 2.9. Redémarrer le kernel

Si votre notebook se comporte bizarrement (variables qui ne correspondent plus à ce que vous attendez, erreurs étranges), le réflexe à avoir est de **redémarrer le Kernel** (*Kernel → Restart Kernel...*) puis de tout exécuter à nouveau depuis le début.


## 3. Notions de base sur Git et GitHub

### 3.1. Définitions

Git est un **système de gestion de versions**. Il permet de conserver l'historique des modifications apportées à un projet et de revenir à une version antérieure si nécessaire. Imaginez que vous travaillez sur un mémoire et que vous enregistriez successivement :

```
memoire_v1.docx
memoire_v2.docx
memoire_v3_final.docx
memoire_v3_final_corrige.docx
memoire_v3_final_corrige_2.docx
```

Git vous aide à conserver l'historique d'un même projet sans multiplier les copies de fichiers.

Quant à GitHub, il s'agit d'une plateforme en ligne qui permet d'héberger des projets utilisant Git. C'est sur GitHub que se trouve le dépôt de ce cours. Ce dernier contient les notebooks, les jeux de données, les exercices et leurs corrigés.

### 3.2. Git ≠ GitHub

Git et GitHub sont liés, mais ce ne sont pas la même chose. Git est un logiciel installé sur votre ordinateur. Il peut fonctionner sans connexion Internet. GitHub est un service en ligne. Il permet notamment de stocker des dépôts Git à distance et de les partager avec d'autres personnes. On peut donc utiliser Git sans GitHub. En revanche, dans ce cours, nous utilisons GitHub comme plateforme pour distribuer les ressources du cours.

```
                 VOTRE ORDINATEUR
              ┌───────────────────┐
              │       Git         │
              │                   │
              │  dépôt local      │
              └─────────┬─────────┘
                        │
                  git pull / push
                        │
                        ▼
                 ┌───────────────┐
                 │    GitHub     │
                 │               │
                 │ dépôt distant │
                 └───────────────┘
```

> **À retenir**
> **Git** → gère l'historique des versions.
> **GitHub** → permet notamment d'héberger et de partager des dépôts Git.
> **Dépôt (repository ou repo)** → dossier du projet suivi par Git.

### 3.3. Installer Git sur votre ordinateur

#### 3.3.1. macOS

Sur macOS, commencez par vérifier si Git n'est pas déjà installé en ouvrant votre terminal et en tapant :

```bash
git --version
```

Si Git est disponible, une version s'affiche. Vous pouvez passer à l'étape suivante. Dans le cas contraire, macOS peut afficher une fenêtre proposant d'installer les *Xcode Command Line Tools*. Acceptez l'installation, puis exécutez à nouveau `git --version`. Si une version de Git s'affiche, l'installation est terminée.

#### 3.3.2 Windows

Rendez-vous sur le site [git-scm.com](https://git-scm.com/install/windows) pour télécharger Git sur Windows, puis installer avec les options pr défaut. Après installation :

```bash
git --version
```

#### 3.3.3. Linux

L'installation dépend de la distribution. Par exemple, sur Ubuntu/Debian :

```bash
sudo apt update
sudo apt install git
```

Puis :

```bash
git --version
```

### 3.4. Télécharger le cours

```bash

```

Cette commande récupère les dernières modifications publiées sur GitHub.

### 3.6. Et si je ne veux pas utiliser Git ?

Ce n'est pas un problème ! Si vous n'êtes pas à l'aise avec le terminal, vous pouvez télécharger le dépôt sous forme de fichier `.zip` depuis GitHub (bouton vert **Code** → **Download ZIP**). Vous pouvez ensuite décompresser le fichier sur votre ordinateur pour accéder aux fichiers du cours.

> Cette méthode est moins pratique pour récupérer les mises à jour, car il faudra télécharger à nouveau le fichier `.zip`, mais elle suffit pour suivre ce cours.

Dans ce module, vous utilisez Git uniquement pour récupérer le cours. Plus tard, dans le module [09-reproductibilite](../09-reproductibilite/), vous apprendrez à utiliser Git pour gérer vos propres projets de recherche. Vous découvrirez alors les notions suivantes :

- `repository` : le projet suivi par Git ;
- `working directory` : les fichiers sur lesquels vous travaillez ;
- `commit` : une sauvegarde documentée d'un ensemble de modifications ;
- `branch` : une version parallèle du projet ;
- `push` : envoyer vos modifications vers GitHub ;
- `pull` : récupérer les modifications depuis GitHub ;
- `clone` : récupérer une copie complète d'un dépôt.
