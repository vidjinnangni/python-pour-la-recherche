# Corrigé Exercice 1 (Module 04) : Statistiques descriptives et anomalies

```python
import pandas as pd
import numpy as np
from scipy import stats

df = pd.read_csv("../../assets/datasets/entreprises_panel.csv")
```

## Partie 1

```python
print("Moyenne :", df["effectifs"].mean())
print("Médiane :", df["effectifs"].median())
print("Écart-type :", df["effectifs"].std())
print("Asymétrie :", df["effectifs"].skew())
```

La moyenne est supérieure à la médiane, et l'asymétrie est positive : la distribution des effectifs est étalée vers la droite (quelques grandes entreprises tirent la moyenne vers le haut), cohérent avec ce qu'on observe généralement pour une variable de taille d'entreprise.


## Partie 2

```python
# 1. Q1, Q3, IQR
q1 = df["chiffre_affaires_keur"].quantile(0.25)
q3 = df["chiffre_affaires_keur"].quantile(0.75)
iqr = q3 - q1
```

```python
# 2. Bornes
borne_basse = q1 - 1.5 * iqr
borne_haute = q3 + 1.5 * iqr
print(borne_basse, borne_haute)
```

```python
# 3. Nombre de lignes atypiques
lignes_atypiques = df[(df["chiffre_affaires_keur"] < borne_basse) | (df["chiffre_affaires_keur"] > borne_haute)]
print(len(lignes_atypiques))
```

## Partie 3

```python
statistique, p_value = stats.shapiro(df["effectifs"])
print(f"Statistique = {statistique:.4f}, p-value = {p_value:.6f}")

if p_value > 0.05:
    print("On ne rejette pas l'hypothèse de normalité")
else:
    print("On rejette l'hypothèse de normalité -- la variable ne semble pas suivre une loi normale")
```
