# Corrigé Exercice 2 (Module 04) : Comparaison de moyennes

```python
import pandas as pd
from scipy import stats

df = pd.read_csv("../../assets/datasets/entreprises_panel.csv")
df["segment"] = df["effectifs"].apply(lambda e: "Petite" if e < 100 else "Grande")
```

## Partie 1

```python
df.groupby("segment")["satisfaction_rh"].mean()
```

## Partie 2

```python
groupe_petite = df[df["segment"] == "Petite"]["satisfaction_rh"].dropna()
groupe_grande = df[df["segment"] == "Grande"]["satisfaction_rh"].dropna()

stat_petite, p_petite = stats.shapiro(groupe_petite)
stat_grande, p_grande = stats.shapiro(groupe_grande)

print(f"Petite entreprise : p-value = {p_petite:.4f}")
print(f"Grande entreprise  : p-value = {p_grande:.4f}")
```

## Partie 3

```python
# Si les deux groupes semblent raisonnablement normaux (p > 0.05 dans les deux cas), le test t est justifié.
# Sinon, Mann-Whitney est le choix plus prudent. Le code ci-dessous illustre les deux -- en pratique,
# on choisirait UN SEUL des deux tests selon le résultat de la partie 2.

stat_t, p_t = stats.ttest_ind(groupe_grande, groupe_petite)
print(f"Test t : statistique = {stat_t:.3f}, p-value = {p_t:.5f}")

stat_mw, p_mw = stats.mannwhitneyu(groupe_grande, groupe_petite)
print(f"Mann-Whitney : statistique = {stat_mw:.1f}, p-value = {p_mw:.5f}")
```

## Partie 4

**Exemple de réponse** : au seuil de 5%, si la p-value obtenue est inférieure à 0.05, on peut dire que l'écart de satisfaction RH observé entre les deux segments n'est probablement pas dû au seul hasard de l'échantillonnage. Il y a une association statistiquement détectable entre la taille de l'entreprise et la satisfaction RH dans cet échantillon.

**Ce qu'on ne peut PAS dire** : on ne peut pas dire que la taille de l'entreprise *cause* ce niveau de satisfaction différent. D'autres facteurs non contrôlés ici (secteur, région, pratiques managériales...) pourraient expliquer une partie ou la totalité de cet écart. Établir un lien causal demanderait une méthode économétrique adaptée (module 05), avec des variables de contrôle.
