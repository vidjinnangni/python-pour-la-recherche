# Corrigé Exercice 4 (Module 04) : Intervalles de confiance

```python
import pandas as pd
import numpy as np
from scipy import stats

df = pd.read_csv("../../assets/datasets/entreprises_panel.csv")

def intervalle_confiance(donnees, niveau=0.95):
    moyenne = donnees.mean()
    erreur_standard = donnees.std() / np.sqrt(len(donnees))
    ic = stats.t.interval(niveau, df=len(donnees) - 1, loc=moyenne, scale=erreur_standard)
    return moyenne, ic
```

## Partie 1

```python
moyenne, ic = intervalle_confiance(df["chiffre_affaires_keur"])
print(f"Moyenne : {moyenne:.2f}")
print(f"IC 95% : [{ic[0]:.2f}, {ic[1]:.2f}]")
```

## Partie 2

```python
donnees_tech = df[df["secteur"] == "Technologies"]["chiffre_affaires_keur"]
moyenne_tech, ic_tech = intervalle_confiance(donnees_tech)
print(f"n = {len(donnees_tech)}")
print(f"Moyenne : {moyenne_tech:.2f}")
print(f"IC 95% : [{ic_tech[0]:.2f}, {ic_tech[1]:.2f}]")
```

L'intervalle est plus **large** pour le sous-échantillon "Technologies" : avec moins d'observations, l'erreur standard (qui dépend de `1 / racine(n)`) est plus élevée, donc l'incertitude sur la vraie moyenne est plus grande — logique, on dispose de moins d'information.
