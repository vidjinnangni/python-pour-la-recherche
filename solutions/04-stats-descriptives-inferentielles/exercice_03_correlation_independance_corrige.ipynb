# Corrigé Exercice 3 (Module 04) : Corrélation et indépendance

```python
import pandas as pd
from scipy import stats

df = pd.read_csv("../../assets/datasets/entreprises_panel.csv")
```

## Partie 1

```python
r, p_value = stats.pearsonr(df["chiffre_affaires_keur"], df["rentabilite"])
print(f"r = {r:.3f}, p-value = {p_value:.5f}")
```

Selon le repère du cours, un coefficient dans cette plage correspond à une force de relation à qualifier selon sa valeur numérique exacte obtenue (négligeable si < 0.1, faible entre 0.1 et 0.3, modérée entre 0.3 et 0.5, forte au-delà de 0.5).


## Partie 2

```python
r_spearman, p_spearman = stats.spearmanr(df["chiffre_affaires_keur"], df["rentabilite"])
print(f"Spearman : r = {r_spearman:.3f}, p-value = {p_spearman:.5f}")
```

Si les deux coefficients (Pearson et Spearman) sont proches, la relation est à la fois linéaire et monotone. Un écart notable entre les deux suggérerait une relation monotone mais non linéaire (ex. en forme de courbe plutôt qu'en ligne droite).


## Partie 3

```python
# 1. Tableau de contingence
tableau = pd.crosstab(df["secteur"], df["teletravail"])
tableau
```

```python
# 2. Test du Chi²
statistique, p_value, ddl, valeurs_attendues = stats.chi2_contingency(tableau)
print(f"Statistique = {statistique:.3f}, p-value = {p_value:.4f}")

if p_value > 0.05:
    print("Indépendance -- pas de lien détecté entre secteur et télétravail")
else:
    print("Les deux variables semblent liées (dépendance détectée)")
```
