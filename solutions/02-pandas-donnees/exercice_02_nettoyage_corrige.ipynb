# Corrigé Exercice 2 (Module 02) : Nettoyage

```python
import pandas as pd

df = pd.read_csv("../../assets/datasets/entreprises_panel.csv")
```

## Partie 1

```python
df.isna().sum()
```

## Partie 2

```python
# 1. Suppression des lignes avec valeur manquante
df_sans_na = df.dropna(subset=["satisfaction_rh"])
print(df_sans_na["satisfaction_rh"].isna().sum())  # doit afficher 0
```

```python
# 2. Remplacement par la médiane
df_rempli = df.copy()
mediane_satisfaction = df_rempli["satisfaction_rh"].median()
df_rempli["satisfaction_rh"] = df_rempli["satisfaction_rh"].fillna(mediane_satisfaction)
print(df_rempli["satisfaction_rh"].isna().sum())  # doit afficher 0
```

## Partie 3

```python
# 1. Nombre de doublons
df.duplicated().sum()
```

```python
# 2. Suppression des doublons
df_propre = df.drop_duplicates()
```

```python
# 3. Nombre de lignes restantes
len(df_propre)
```

## Partie 4

```python
# 1. Type de la colonne effectifs
df["effectifs"].dtype
```

```python
# 2. Conversion "de sécurité" et vérification
effectifs_convertis = pd.to_numeric(df["effectifs"], errors="coerce")
print(effectifs_convertis.isna().sum())  # 0 attendu : aucune valeur n'a été perdue dans la conversion
```
