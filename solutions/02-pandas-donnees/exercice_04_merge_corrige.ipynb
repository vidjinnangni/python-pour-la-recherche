# Corrigé Exercice 4 (Module 02) : Croiser deux jeux de données

```python
import pandas as pd

df = pd.read_csv("../../assets/datasets/entreprises_panel.csv")

certifications = pd.DataFrame({
    "entreprise_id": ["E001", "E003", "E007", "E012", "E020", "E035", "E050"],
    "certification_iso": [True, True, False, True, True, False, True],
})
```

## Partie 1

**Réponse attendue** : avec `how="left"`, on garde toutes les lignes de `df` (404 lignes, comme `df` lui-même), puisque `df` est le DataFrame de gauche. Avec `how="inner"`, on ne garde que les lignes dont l'`entreprise_id` existe aussi dans `certifications` ; uniquement les lignes concernant les 7 entreprises listées dans `certifications`, sur les 5 années de panel chacune, soit au maximum 35 lignes (un peu moins si certaines années manquent pour ces entreprises).


```python
print(len(df.merge(certifications, on="entreprise_id", how="left")))
print(len(df.merge(certifications, on="entreprise_id", how="inner")))
```

## Partie 2

```python
df_avec_certif = df.merge(certifications, on="entreprise_id", how="left")
df_avec_certif["certification_iso"].isna().sum()
```

**Explication** : ces valeurs manquantes correspondent aux entreprises qui ne figurent pas dans `certifications`. On n'a simplement pas d'information sur leur certification. Pandas met donc `NaN` plutôt que d'inventer une valeur.


## Partie 3

```python
df_certif_seulement = df.merge(certifications, on="entreprise_id", how="inner")
print(len(df_certif_seulement))                                                 # nombre de lignes
print(df_certif_seulement["entreprise_id"].nunique())           # nombre d'entreprises différentes
```

## Partie 4

```python
df_avec_certif.groupby("certification_iso")["chiffre_affaires_keur"].mean()
```

**Observation** : les entreprises dont `certification_iso` est `NaN` **n'apparaissent pas** dans ce résultat. `groupby` ignore par défaut les groupes `NaN`. C'est un comportement à connaître. Si vous voulez explicitement inclure les valeurs manquantes comme un groupe à part, il faut utiliser l'option `dropna=False` de `groupby()`.
