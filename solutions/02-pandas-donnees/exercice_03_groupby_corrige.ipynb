# Corrigé Exercice 3 (Module 02) : Groupby et statistiques par groupe

```python
import pandas as pd

df = pd.read_csv("../../assets/datasets/entreprises_panel.csv")
```

## Partie 1

```python
df.groupby("secteur")["rentabilite"].mean().sort_values(ascending=False)
```

Le secteur en tête de ce tri est celui avec la rentabilité moyenne la plus élevée.

## Partie 2

```python
df.groupby("secteur")["chiffre_affaires_keur"].agg(["mean", "std", "count"])
```

## Partie 3

```python
df.groupby(["secteur", "teletravail"])["effectifs"].mean()
```

## Partie 4

```python
pd.pivot_table(
    df,
    values="chiffre_affaires_keur",
    index="secteur",
    columns="teletravail",
    aggfunc="mean",
)
```

L'écart entre les colonnes `0` et `1` varie selon les secteurs. C'est une première observation visuelle, mais rappelez-vous. À ce stade, on ne sait pas encore si ces écarts sont statistiquement significatifs (module 04) ni s'ils reflètent un vrai effet causal du télétravail plutôt que d'autres facteurs (module 05).
