# Corrigé Exercice 2 (Module 03) : Personnaliser un graphique

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("../../assets/datasets/entreprises_panel.csv")

evolution = df.groupby("annee")["chiffre_affaires_keur"].mean()
```

## Partie 1

Problèmes typiques du graphique de départ :
1. Pas de titre
2. Pas de labels sur les axes (on ne sait pas ce que représentent x et y)
3. Figure trop petite / taille par défaut peu lisible
4. Pas de marqueurs sur les points, difficile de voir les années individuelles
5. (bonus) L'axe Y ne part pas de 0, ce qui peut exagérer visuellement la variation


## Partie 2

```python
fig, ax = plt.subplots(figsize=(8, 5))

ax.plot(evolution.index, evolution.values, marker="o")

ax.set_xlabel("Année")
ax.set_ylabel("Chiffre d'affaires moyen (k€)")
ax.set_title("Évolution du chiffre d'affaires moyen des entreprises (2019-2023)")

ax.set_xticks(evolution.index)

plt.show()
```

## Partie 3

```python
fig, ax = plt.subplots(figsize=(8, 5))

ax.plot(evolution.index, evolution.values, marker="o")

ax.set_xlabel("Année")
ax.set_ylabel("Chiffre d'affaires moyen (k€)")
ax.set_title("Évolution du chiffre d'affaires moyen des entreprises (2019-2023)")

ax.set_ylim(0, evolution.max() * 1.1)  # 10% de marge au-dessus du maximum, pour la lisibilité

ax.set_xticks(evolution.index)

plt.show()
```

**Observation** : sans `set_ylim(0, ...)`, `matplotlib` zoome automatiquement sur la plage de valeurs observées ; ce qui peut donner l'impression d'une variation beaucoup plus marquée qu'elle ne l'est réellement en proportion. Avec l'axe Y forcé à partir de 0, la variation apparaît pour ce qu'elle est réellement. C'est un choix éditorial à adapter selon ce que vous voulez réellement montrer, mais toujours en connaissance de cause plutôt que par défaut.
