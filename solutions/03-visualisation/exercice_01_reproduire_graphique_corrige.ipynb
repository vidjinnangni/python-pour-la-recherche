# Corrigé Exercice 1 (Module 03) : Reproduire un graphique donné

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("../../assets/datasets/entreprises_panel.csv")
```

## Partie 1

- **Demande A** (répartition d'une variable continue) → **histogramme**
- **Demande B** (comparer un compte par catégorie) → **barres**
- **Demande C** (relation entre deux variables numériques) → **nuage de points**


## Partie 2

```python
# Graphique A -- histogramme des effectifs

fig, ax = plt.subplots(figsize=(7, 4))

ax.hist(df["effectifs"], bins=20, edgecolor="white")

ax.set_xlabel("Effectifs")
ax.set_ylabel("Nombre d'observations")
ax.set_title("Distribution des effectifs des entreprises")

plt.show()
```

```python
# Graphique B -- nombre d'entreprises par secteur

compte_secteur = df["secteur"].value_counts()

fig, ax = plt.subplots(figsize=(7, 4))

ax.bar(compte_secteur.index, compte_secteur.values)

ax.set_xlabel("Secteur")
ax.set_ylabel("Nombre d'observations")
ax.set_title("Nombre d'observations par secteur")

plt.xticks(rotation=30, ha="right")

plt.show()
```

```python
# Graphique C -- rentabilité vs satisfaction RH

fig, ax = plt.subplots(figsize=(7, 4))

ax.scatter(df["rentabilite"], df["satisfaction_rh"], alpha=0.6)

ax.set_xlabel("Rentabilité")
ax.set_ylabel("Satisfaction RH")
ax.set_title("Rentabilité en fonction de la satisfaction RH")

plt.show()
```
