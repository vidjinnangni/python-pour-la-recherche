# Corrigé Exercice 4 (Module 03) : Graphique interactif avec plotly

```python
import pandas as pd
import plotly.express as px

df = pd.read_csv("../../assets/datasets/entreprises_panel.csv")
```

## Partie 1

```python
ca_par_secteur = df.groupby("secteur")["chiffre_affaires_keur"].mean().reset_index()

fig = px.bar(
    ca_par_secteur,
    x="secteur",
    y="chiffre_affaires_keur",
    title="Chiffre d'affaires moyen par secteur",
)

fig.show()
```

## Partie 2

```python
fig = px.scatter(
    df,
    x="effectifs",
    y="rentabilite",
    color="teletravail",
    hover_data=["entreprise_id", "annee", "secteur"],
    title="Rentabilité en fonction des effectifs, selon le télétravail",
)

fig.show()
```

## Partie 3

```python
satisfaction_annee = df.groupby(["annee", "teletravail"])["satisfaction_rh"].mean().reset_index()

fig = px.line(
    satisfaction_annee,
    x="annee",
    y="satisfaction_rh",
    color="teletravail",
    title="Satisfaction RH moyenne par année, selon le télétravail",
)
fig.show()
```
