# Corrigé Exercice 3 (Module 03) : Corrélation avec seaborn

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("../../assets/datasets/entreprises_panel.csv")
```

## Partie 1

```python
fig, ax = plt.subplots(figsize=(7, 5))

sns.scatterplot(data=df, x="effectifs", y="chiffre_affaires_keur", ax=ax)

ax.set_xlabel("Effectifs")
ax.set_ylabel("Chiffre d'affaires (k€)")
ax.set_title("Chiffre d'affaires en fonction des effectifs")

plt.show()
```

## Partie 2

```python
fig, ax = plt.subplots(figsize=(7, 5))

sns.scatterplot(data=df, x="effectifs", y="chiffre_affaires_keur", hue="secteur", ax=ax)

ax.set_xlabel("Effectifs")
ax.set_ylabel("Chiffre d'affaires (k€)")
ax.set_title("Chiffre d'affaires en fonction des effectifs, par secteur")

plt.show()
```

Avec la **couleur**, on voit si **la relation effectifs/chiffre d'affaires** est similaire d'un secteur à l'autre ou si certains secteurs se distinguent (ex. un secteur avec un chiffre d'affaires par salarié structurellement plus élevé, qui se traduirait par un nuage de points "au-dessus" des autres). Sans cette couleur, tous les points se mélangent et cette distinction sectorielle sera invisible.


## Partie 3

```python
# 1. Matrice de corrélation
correlations = df[["effectifs", "chiffre_affaires_keur", "rentabilite", "satisfaction_rh"]].corr()
correlations
```

```python
# 2. Heatmap
fig, ax = plt.subplots(figsize=(6, 5))

sns.heatmap(correlations, annot=True, cmap="coolwarm", center=0, ax=ax)

ax.set_title("Corrélations entre variables numériques")

plt.show()
```

**Réponse à la question 3** : la corrélation la plus forte (en valeur absolue, en excluant la diagonale qui vaut toujours 1) est celle entre `effectifs` et `chiffre_affaires_keur`. Ce qui est cohérent : plus une entreprise est grande, plus son chiffre d'affaires tend à être élevé, une relation assez mécanique en réalité.
