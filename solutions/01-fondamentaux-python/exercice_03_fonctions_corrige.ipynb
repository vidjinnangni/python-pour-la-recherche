# Corrigé Exercice 3 (Module 01) : Fonctions

## Partie 1

```python
def moyenne(valeurs):
    return sum(valeurs) / len(valeurs)

print(moyenne([1, 2, 3, 4, 5]))  # 3.0
```

## Partie 2

```python
def moyenne_ponderee(valeurs, poids):
    total_pondere = sum(v * p for v, p in zip(valeurs, poids))
    total_poids = sum(poids)
    return total_pondere / total_poids

print(moyenne_ponderee([1, 2, 3, 4, 5], [1, 1, 1, 1, 1]))  # 3.0
print(moyenne_ponderee([4, 5, 2, 3], [1, 2, 1, 1]))         # 3.8
```

## Partie 3

```python
# Sans protection, ceci lève une ZeroDivisionError :
try:
    moyenne_ponderee([], [])
except ZeroDivisionError as e:
    print("Erreur obtenue :", e)
```

```python
def moyenne_ponderee(valeurs, poids):
    if not poids or sum(poids) == 0:
        return None
    total_pondere = sum(v * p for v, p in zip(valeurs, poids))
    total_poids = sum(poids)
    return total_pondere / total_poids

print(moyenne_ponderee([], []))                       # None
print(moyenne_ponderee([4, 5, 2, 3], [1, 2, 1, 1]))    # 3.8
```

## Partie 4

```python
repondants = [
    {"satisfaction": 4, "anciennete": 5.5},
    {"satisfaction": 5, "anciennete": 2.0},
    {"satisfaction": 2, "anciennete": 12.0},
    {"satisfaction": 3, "anciennete": 20.0},
]

satisfactions = [r["satisfaction"] for r in repondants]
anciennetes = [r["anciennete"] for r in repondants]

resultat = moyenne_ponderee(satisfactions, anciennetes)
print(f"Satisfaction moyenne pondérée : {resultat:.2f}")
```
