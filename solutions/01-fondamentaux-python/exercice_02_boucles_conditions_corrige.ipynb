# Corrigé Exercice 2 (Module 01) : Boucles et conditions

```python
effectifs_entreprises = [8, 45, 230, 12, 95, 340, 28, 5, 180, 60]
```

## Partie 1

```python
pme = []
for effectif in effectifs_entreprises:
    if effectif < 250:
        pme.append(effectif)

print(pme)
```

## Partie 2

```python
pme_v2 = [e for e in effectifs_entreprises if e < 250]
print(pme_v2)
print(pme == pme_v2)
```

## Partie 3

```python
for effectif in effectifs_entreprises:
    if effectif < 10:
        print(f"{effectif} -> Micro-entreprise")
    elif effectif < 50:
        print(f"{effectif} -> Petite entreprise")
    elif effectif < 250:
        print(f"{effectif} -> Moyenne entreprise")
    else:
        print(f"{effectif} -> Grande entreprise")
```

## Partie 4

```python
compteur = {"Micro-entreprise": 0, "Petite entreprise": 0, "Moyenne entreprise": 0, "Grande entreprise": 0}

for effectif in effectifs_entreprises:
    if effectif < 10:
        compteur["Micro-entreprise"] += 1
    elif effectif < 50:
        compteur["Petite entreprise"] += 1
    elif effectif < 250:
        compteur["Moyenne entreprise"] += 1
    else:
        compteur["Grande entreprise"] += 1

print(compteur)
```
